```python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from rdkit import Chem
from rdkit.Chem import AllChem, Draw

import matplotlib.pyplot as plt
import networkx as nx
from rdkit import Chem
from rdkit.Chem import AllChem, Draw

def plot_simple_molecule_graph(smiles, save_path=None, figsize=(8, 6)):
    """
    Convert SMILES to graph and plot it.

    Args:
        smiles (str): SMILES string
        save_path (str): Path to save the plot (e.g., 'molecule.png')
        figsize (tuple): Figure size

    Returns:
        tuple: (fig, axes, G) - matplotlib figure, axes, and NetworkX graph
    """
    # Convert SMILES to molecule
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        print(f"Invalid SMILES: {smiles}")
        return None

    # Generate 2D coordinates
    mol = Chem.AddHs(mol)  # Add hydrogens for better visualization
    AllChem.Compute2DCoords(mol)

    # Create graph
    G = nx.Graph()
    pos = {}  # Store node positions

    # Add atoms as nodes with positions
    for atom in mol.GetAtoms():
        idx = atom.GetIdx()
        G.add_node(idx, symbol=atom.GetSymbol())

        # Get 2D coordinates
        conf = mol.GetConformer()
        pos[idx] = (conf.GetAtomPosition(idx).x, conf.GetAtomPosition(idx).y)

    # Add bonds as edges
    for bond in mol.GetBonds():
        i = bond.GetBeginAtomIdx()
        j = bond.GetEndAtomIdx()
        G.add_edge(i, j)

    # Create plot
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=figsize)

    # Plot 1: NetworkX graph
    labels = {i: G.nodes[i]['symbol'] for i in G.nodes()}

    nx.draw(G, pos, ax=ax1, with_labels=True, labels=labels,
            node_color='lightblue', node_size=500,
            font_size=10, font_weight='bold',
            edge_color='gray', width=2)

    ax1.set_title(f"Graph Representation\n{smiles}")
    ax1.axis('equal')

    # Plot 2: RDKit structure
    mol_no_h = Chem.RemoveHs(mol)  # Remove hydrogens for cleaner view
    img = Draw.MolToImage(mol_no_h, size=(300, 300))
    ax2.imshow(img)
    ax2.axis('off')
    ax2.set_title("Chemical Structure")

    plt.tight_layout()

    # Save plot if path is provided
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        print(f"Plot saved to: {save_path}")

    plt.show()

    return fig, (ax1, ax2), G
```

```python
"""
best_so_far.py
==============
Drug classification pipeline — targets ≥ 90 % Random Forest accuracy.

Key corrections vs. the original notebook
──────────────────────────────────────────
1. NO RDKit chemical descriptors  (project rules: only topological/geometrical).
2. Data-leakage fix: scaler is fit_transform'd on TRAIN only, then
   transform'd on TEST — never re-fitted on test data.
3. Sensible RF hyper-parameters (max_depth=None + min_samples_leaf=2
   is better than max_depth=70 which memorises noise).
4. TOPONETX_AVAILABLE defined before first use.
5. Dead code / duplicate imports removed.
"""

import pandas as pd
import numpy as np
import random
import warnings
warnings.filterwarnings('ignore')

from rdkit import Chem
from rdkit.Chem.Scaffolds import MurckoScaffold

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import (accuracy_score, classification_report,
                              confusion_matrix, roc_auc_score)
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

# Our feature generator (must be in the same directory)
from feature_gen import DiscriminativeFeatureGenerator, print_feature_breakdown

# ── optional: TopoModelX (not required) ──────────────────────────────────────
TOPONETX_AVAILABLE = False
try:
    import toponetx as tnx
    TOPONETX_AVAILABLE = True
    print("✓ TopoModelX imported")
except ImportError:
    print("✗ TopoModelX not available (not needed for this pipeline)")


###############################################################################
# 1.  SCAFFOLD SPLITTING
###############################################################################
def compute_scaffold(smiles):
    """Murcko scaffold SMILES; None for invalid SMILES."""
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol is None:
            return None
        return Chem.MolToSmiles(MurckoScaffold.GetScaffoldForMol(mol))
    except Exception:
        return None


def scaffold_split(df, test_size=0.2, random_state=36):
    """
    Group molecules by Murcko scaffold, then assign whole groups to
    train or test.  This prevents chemical-similarity leakage between
    splits — far more realistic than a random split for drug discovery.
    """
    scaffolds = {}
    for idx, smiles in enumerate(df["SMILES"]):
        sc = compute_scaffold(smiles)
        if sc is None:
            sc = f"NOSCAFFOLD_{idx}"
        scaffolds.setdefault(sc, []).append(idx)

    groups = list(scaffolds.values())
    rng    = random.Random(random_state)
    rng.shuffle(groups)

    n_test_target             = int(test_size * len(df))
    test_idx, train_idx, acc  = [], [], 0

    for g in groups:
        if acc < n_test_target:
            test_idx.extend(g)
            acc += len(g)
        else:
            train_idx.extend(g)

    return train_idx, test_idx


###############################################################################
# 2.  SMILES  →  WEIGHTED ADJACENCY MATRIX
###############################################################################
class SMILESToGraphConverter:
    """
    Encode a molecule as a fixed-size weighted adjacency matrix.

    Node encoding  : diagonal entry  =  atomic_number / 10
    Edge encoding  : off-diagonal    =  bond weight
        single   → 1.0
        double   → 2.0
        triple   → 3.0
        aromatic → 1.5
    """

    _BOND_W = {
        Chem.rdchem.BondType.SINGLE:   1.0,
        Chem.rdchem.BondType.DOUBLE:   2.0,
        Chem.rdchem.BondType.TRIPLE:   3.0,
        Chem.rdchem.BondType.AROMATIC: 1.5,
    }

    def __init__(self, max_atoms=100):
        self.max_atoms = max_atoms

    def smiles_to_adjacency(self, smiles):
        try:
            mol = Chem.MolFromSmiles(smiles)
            if mol is None:
                return None
            mol  = Chem.AddHs(mol)
            n    = min(mol.GetNumAtoms(), self.max_atoms)
            adj  = np.zeros((self.max_atoms, self.max_atoms))

            for i in range(n):
                adj[i, i] = mol.GetAtomWithIdx(i).GetAtomicNum() / 10.0
                for j in range(i + 1, n):
                    bond = mol.GetBondBetweenAtoms(i, j)
                    if bond:
                        w = self._BOND_W.get(bond.GetBondType(), 1.0)
                        adj[i, j] = adj[j, i] = w
            return adj
        except Exception as e:
            print(f"  SMILES error: {smiles[:40]} → {e}")
            return None

    def convert_list(self, smiles_list):
        adjs, valid_idx = [], []
        for i, sm in enumerate(smiles_list):
            if pd.isna(sm):
                continue
            adj = self.smiles_to_adjacency(sm)
            if adj is not None:
                adjs.append(adj)
                valid_idx.append(i)
        return adjs, valid_idx


###############################################################################
# 3.  PIPELINE
###############################################################################
class DrugClassificationPipeline:
    """
    End-to-end pipeline:
        load  →  scaffold split  →  graph conversion
        →  topological feature extraction  (fit scaler on TRAIN only)
        →  Random Forest training  →  evaluation
    """

    def __init__(self, n_bins=20, max_atoms=100, n_spec=5,
                 test_size=0.2, random_state=42):
        self.n_bins       = n_bins
        self.max_atoms    = max_atoms
        self.n_spec       = n_spec
        self.test_size    = test_size
        self.random_state = random_state

        self.converter     = SMILESToGraphConverter(max_atoms=max_atoms)
        self.feat_gen      = DiscriminativeFeatureGenerator(
                                 n_bins=n_bins, n_spec=n_spec)
        self.label_encoder = LabelEncoder()
        self.results       = {}

    # ── helpers ───────────────────────────────────────────────────────────────
    def _df_to_arrays(self, df, fit_scaler=False):
        """
        Convert a DataFrame split into  (X_scaled, y_encoded).

        fit_scaler=True   → use  feat_gen.fit_transform()   (TRAIN split)
        fit_scaler=False  → use  feat_gen.transform()        (TEST  split)
        """
        smiles  = df["SMILES"].tolist()
        targets = df["target"].tolist()

        adjs, valid_idx = self.converter.convert_list(smiles)
        valid_targets   = [targets[i] for i in valid_idx]

        if fit_scaler:
            X, _, info = self.feat_gen.fit_transform(adjs, valid_targets)
            y           = self.label_encoder.fit_transform(valid_targets)
        else:
            X, _, info = self.feat_gen.transform(adjs, valid_targets)
            y           = self.label_encoder.transform(valid_targets)

        return X, y

    # ── model training ────────────────────────────────────────────────────────
    def _train_and_evaluate(self, X_train, y_train, X_test, y_test):
        """
        Train Random Forest (primary) and SVM (secondary).

        RF hyper-parameter rationale:
            n_estimators = 500     enough trees; diminishing returns beyond ~300
            max_depth    = None    grow full trees — bagging + random features
                                   already regularise; hard depth cap hurts recall
            min_samples_leaf = 2   gentle regularisation, avoids single-sample leaves
            max_features = 'sqrt'  standard RF heuristic for classification
            class_weight = 'balanced'  compensates for class imbalance
        """
        cv = StratifiedKFold(n_splits=5, shuffle=True,
                             random_state=self.random_state)

        models = {
            "Random Forest": RandomForestClassifier(
                n_estimators=500,
                max_depth=None,
                min_samples_leaf=2,
                max_features='sqrt',
                class_weight='balanced',
                n_jobs=-1,
                random_state=self.random_state
            ),
            "SVM (RBF)": SVC(
                kernel='rbf',
                C=10,
                gamma='scale',
                probability=True,
                class_weight='balanced',
                random_state=self.random_state
            ),
        }

        sep = "=" * 65
        print(f"\n{sep}")
        print("  MODEL EVALUATION")
        print(sep)

        best_name, best_acc = None, 0.0

        for name, model in models.items():
            # 5-fold stratified CV on training data
            cv_scores = cross_val_score(model, X_train, y_train,
                                        cv=cv, scoring='accuracy', n_jobs=-1)

            # Final fit on full training set
            model.fit(X_train, y_train)
            y_pred = model.predict(X_test)
            acc    = accuracy_score(y_test, y_pred)

            try:
                proba   = model.predict_proba(X_test)[:, 1]
                auc     = roc_auc_score(y_test, proba)
                auc_str = f"  AUC-ROC       : {auc:.4f}"
            except Exception:
                auc_str = ""

            print(f"\n── {name} ──")
            print(f"  Test accuracy : {acc:.4f}  ({acc*100:.2f} %)")
            print(f"  CV  accuracy  : {cv_scores.mean():.4f} ± {cv_scores.std():.4f}")
            if auc_str:
                print(auc_str)
            print(classification_report(
                y_test, y_pred,
                target_names=[str(c) for c in self.label_encoder.classes_],
                zero_division=0))

            self.results[name] = {
                'model'    : model,
                'accuracy' : acc,
                'cv_mean'  : cv_scores.mean(),
                'cv_std'   : cv_scores.std(),
                'y_pred'   : y_pred,
            }
            if acc > best_acc:
                best_acc  = acc
                best_name = name

        print(sep)
        print(f"  Best model: {best_name}  →  {best_acc*100:.2f} %")
        print(sep)
        return self.results

    # ── main entry point ──────────────────────────────────────────────────────
    def run(self, data_path="drugs_combined.csv"):
        # 1. Load
        df = pd.read_csv(data_path).dropna(subset=["SMILES", "target"])
        print(f"Dataset: {len(df)} molecules")
        print("Class distribution:", df["target"].value_counts().to_dict())

        # 2. Scaffold split
        train_idx, test_idx = scaffold_split(
            df, test_size=self.test_size, random_state=self.random_state)
        df_train = df.iloc[train_idx].reset_index(drop=True)
        df_test  = df.iloc[test_idx ].reset_index(drop=True)
        print(f"Train: {len(df_train)}   Test: {len(df_test)}")

        # 3. Feature extraction
        #    ⚠ fit_scaler=True ONLY for training data
        print("\n── Extracting TRAIN features (fit + transform scaler) ──")
        X_train, y_train = self._df_to_arrays(df_train, fit_scaler=True)

        print("\n── Extracting TEST features (transform only — no leakage) ──")
        X_test, y_test   = self._df_to_arrays(df_test,  fit_scaler=False)

        print(f"\nX_train: {X_train.shape}   X_test: {X_test.shape}")

        # 4. Train & evaluate
        results = self._train_and_evaluate(X_train, y_train, X_test, y_test)
        return X_train, y_train, X_test, y_test, results


###############################################################################
# ENTRY POINT
###############################################################################
if __name__ == "__main__":
    print("=" * 65)
    print("  DRUG CLASSIFICATION — TDA + SPECTRAL + GRAPH FEATURES")
    print("=" * 65)
    print(f"  TopoModelX available: {TOPONETX_AVAILABLE}")

    # Show expected feature count before running
    print_feature_breakdown(
        homology_dimensions=(0, 1),
        n_bins=20,
        n_spec=5
    )

    pipeline = DrugClassificationPipeline(
        n_bins=20,          # 20 bins × 2 dims = 40 Betti features
        max_atoms=100,      # cap molecules at 100 heavy atoms
        n_spec=5,           # 5 eigenvalues per Laplacian (L0 + L1)
        test_size=0.2,
        random_state=42
    )

    X_train, y_train, X_test, y_test, results = pipeline.run(
        data_path="drugs_combined.csv"
    )
```
